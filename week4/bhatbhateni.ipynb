## Handling Missing Values ##

```python
df.head()
```

```python
# filling the customer name as missing
df["CustomerName"]=df["CustomerName"].fillna('missing')
```

```python
# finding the most repetitive product category
df["ProductCategory"].describe(include="all")
```

```python
most_frequent = df['ProductCategory'].value_counts().index[0]
most_frequent
```

```python
# filling the missing product category by the most repetitive purchase
df["ProductCategory"]=df["ProductCategory"].fillna('Grocery')
```

```python
df["UnitPrice"] = df["UnitPrice"].fillna(
    df["TotalAmount"] / df["Quantity"]
)
```

```python
df['PaymentMethod'] = df['PaymentMethod'].fillna('Unknown')
```

```python
print(df.isnull().sum())
```

## Formatting the Date Time Column ##

```python
df = df.drop_duplicates().reset_index(drop=True)
print(f"Rows after removing duplicates: {len(df)}")
```

## Formatting the Date Column ##

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['DayOfWeek'] = df['Date'].dt.dayofweek
df['MonthName'] = df['Date'].dt.strftime('%B')
print(df[['Date', 'Year', 'Month', 'Day', 'DayOfWeek', 'MonthName']].head())
```

## Splitting Branch into City ##

```python
df['City'] = df['Branch'].str.split(' - ').str[0]
print(df[['Branch', 'City']].head())
```

## Validating TotalAmount ##

```python
df['CalculatedTotal'] = df['Quantity'] * df['UnitPrice']
df['TotalMatch'] = df['TotalAmount'] == df['CalculatedTotal']
print(df['TotalMatch'].value_counts())
mismatches = df[df['TotalMatch'] == False]
print(f"Mismatches: {len(mismatches)}")
```

## Product Category Distribution ##

```python
category_counts = df['ProductCategory'].value_counts()
print(category_counts)
```

## Branch Distribution ##

```python
branch_counts = df['Branch'].value_counts()
print(branch_counts)
```

## Most Common Payment Method ##

```python
payment_counts = df['PaymentMethod'].value_counts()
print(payment_counts)
print(f"Most common: {payment_counts.index[0]}")
```

## TotalAmount Distribution ##

```python
print(df['TotalAmount'].describe())
skewness = df['TotalAmount'].skew()
print(f"Skewness: {skewness:.2f}")
if skewness > 1:
    print("The distribution is highly skewed to the right.")
elif skewness > 0.5:
    print("The distribution is moderately skewed to the right.")
else:
    print("The distribution is roughly symmetric.")
```

## Monthly Revenue Trend ##

```python
monthly_revenue = df.groupby(['Year', 'Month'])['TotalAmount'].sum().reset_index()
monthly_revenue['YearMonth'] = monthly_revenue['Year'].astype(str) + '-' + monthly_revenue['Month'].astype(str).str.zfill(2)
print(monthly_revenue[['YearMonth', 'TotalAmount']])
```

## Weekend vs Weekday Sales ##

```python
df['IsWeekend'] = df['DayOfWeek'].isin([5, 6])
weekend_sales = df.groupby('IsWeekend')['TotalAmount'].sum()
print(weekend_sales)
print(f"\nWeekend sales: {weekend_sales.get(True, 0):.2f}")
print(f"Weekday sales: {weekend_sales.get(False, 0):.2f}")
```

## Revenue by Day of Week ##

```python
day_names = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
daily_revenue = df.groupby('DayOfWeek')['TotalAmount'].sum().reindex(range(7))
daily_revenue.index = daily_revenue.index.map(lambda x: day_names[x])
print(daily_revenue)
print(f"\nHighest revenue day: {daily_revenue.idxmax()} ({daily_revenue.max():.2f})")
```

## Branch Revenue ##

```python
branch_revenue = df.groupby('Branch')['TotalAmount'].sum().sort_values(ascending=False)
print(branch_revenue)
print(f"\nHighest revenue branch: {branch_revenue.index[0]}")
```

## Average Transaction Value by Branch ##

```python
branch_avg = df.groupby('Branch')['TotalAmount'].mean().sort_values(ascending=False)
print(branch_avg)
```

## City Revenue ##

```python
city_revenue = df.groupby('City')['TotalAmount'].sum().sort_values(ascending=False)
print(city_revenue)
print(f"\nHighest revenue city: {city_revenue.index[0]}")
```

## Category Revenue vs Transactions ##

```python
category_stats = df.groupby('ProductCategory').agg(
    TotalRevenue=('TotalAmount', 'sum'),
    TotalTransactions=('TransactionID', 'count')
).sort_values('TotalRevenue', ascending=False)
print(category_stats)
```

## Top 10 Products by Quantity ##

```python
top_qty = df.groupby('ProductName')['Quantity'].sum().sort_values(ascending=False).head(10)
print(top_qty)
```

## Top 10 Products by Revenue ##

```python
top_revenue = df.groupby('ProductName')['TotalAmount'].sum().sort_values(ascending=False).head(10)
print(top_revenue)
```

## Top 10 Customers by Spend ##

```python
top_customers = df.groupby('CustomerName')['TotalAmount'].sum().sort_values(ascending=False).head(10)
print(top_customers)
```

## Repeat vs One-Time Customers ##

```python
customer_visits = df.groupby('CustomerID')['TransactionID'].nunique()
repeat_customers = (customer_visits > 1).sum()
one_time_customers = (customer_visits == 1).sum()
print(f"Repeat customers: {repeat_customers}")
print(f"One-time customers: {one_time_customers}")
```

## Average Spend per Customer ##

```python
customer_spend = df.groupby('CustomerID')['TotalAmount'].sum()
avg_spend = customer_spend.mean()
print(f"Average spend per customer: {avg_spend:.2f}")
```

## Payment Method by Branch ##

```python
payment_by_branch = pd.crosstab(df['Branch'], df['PaymentMethod'], normalize='index') * 100
print(payment_by_branch.round(1))
```

## Avg Transaction Value by Payment Method ##

```python
avg_by_payment = df.groupby('PaymentMethod')['TotalAmount'].mean().sort_values(ascending=False)
print(avg_by_payment)
```

## Correlation: Quantity, UnitPrice, TotalAmount ##

```python
correlation = df[['Quantity', 'UnitPrice', 'TotalAmount']].corr()
print(correlation)
```

## Outliers in TotalAmount ##

```python
Q1 = df['TotalAmount'].quantile(0.25)
Q3 = df['TotalAmount'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = df[(df['TotalAmount'] < lower_bound) | (df['TotalAmount'] > upper_bound)]
print(f"Number of outliers: {len(outliers)}")
print(f"Lower bound: {lower_bound:.2f}")
print(f"Upper bound: {upper_bound:.2f}")
```
